```python
Step 1: Import NumPy
```

```python
import numpy as np
```

```python
Step 2: Create Employee Data
```

```python
performance = np.array([
    [75, 80, 78, 82],
    [60, 65, 70, 68],
    [90, 92, 88, 94],
    [45, 50, 48, 55]
])
```

```python
Base Salaries
```

```python
salaries = np.array([30000, 28000, 50000, 25000])
```

```python
Step 3: Average Performance
```

```python
avg_performance = np.mean(performance, axis=1)
```

```python
Step 4: Performance Rating
```

```python
ratings = np.where(
    avg_performance >= 85, "Excellent",
    np.where(avg_performance >= 70, "Good",
    np.where(avg_performance >= 50, "Average", "Poor"))
)
```

```python
Step 5: Salary Increment Logic 
```

```python
increments = np.where(
    avg_performance >= 85, 0.20,
    np.where(avg_performance >= 70, 0.10,
    np.where(avg_performance >= 50, 0.05, 0.00))
)

new_salaries = salaries + (salaries * increments)
```

```python
Step 6: Top & Low Performers
```

```python
top_employee = np.argmax(avg_performance)
low_employee = np.argmin(avg_performance)
```

```python
Step 7: Final Report Output
```

```python
for i in range(len(salaries)):
    print("Employee", i + 1)
    print("Average Performance:", avg_performance[i])
    print("Rating:", ratings[i])
    print("Old Salary:", salaries[i])
    print("New Salary:", new_salaries[i])
    print("------------------")
```
